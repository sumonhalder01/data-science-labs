##  1. Data Analysis with Pandas
* **DataFrame:** A 2D labeled data structure with columns of potentially different types (like a SQL table or Excel sheet).
* **Core Operations:**
  * `pd.read_csv()`: Loads external comma-separated data.
  * `.head()` / `.tail()`: Previews data rows.
  * `.iloc[]`: Selects data points by integer location/index position.
  * `.loc[]`: Selects data points by label names.


### 2. Inspecting and Preprocessing Structural Records
* `df.head(n)`: Previews the top `n` rows of a large dataset.
* `df.tail(n)`: Previews the final `n` rows of the file.
* `df.info()`: Reveals column data types, memory footprints, and counts missing entries.
* `df.describe()`: Displays key descriptive stats (Mean, Median, Std Dev, Max, Min).


```python
import pandas as pd

# 1. Building a single dimensional Series from a list
marks_list = [85, 92, 78, 90]
marks_series = pd.Series(marks_list, name="Exam_Scores")
print("--- Pandas Series Structure ---")
print(marks_series)

# 2. Ingesting raw dictionary data into a 2D labeled DataFrame
enterprise_data = {
    "Employee": ["Sumon", "Amit", "Rahul", "Priya", "Ankit"],
    "Role": ["Data Scientist", "Data Analyst", "ML Engineer", "BI Consultant", "Intern"],
    "Salary_INR_K": [95, 90, 80, 65, 45],
    "Remote_Worker": [True, False, True, True, False]
}
df = pd.DataFrame(enterprise_data)

print("\n--- Structural DataFrame Layout ---")
print(df)
```

```python
print("--- Displaying the Top 2 Records ---")
print(df.head(2))

print("\n--- Summary Diagnostic Overview ---")
df.info()

print("\n--- Descriptive Analytical Summary Statistics ---")
print(df.describe())
```

### 3. Precision Ingestion: Advanced Indexing & Column Filtering
* **Isolating Columns:** Passing an array string inside square brackets isolates targeted metrics.
* **`df.iloc[]` (Integer Location):** Selects exact coordinate cells using numeric index coordinates `[Row_Index, Column_Index]`.
* **`df.loc[]` (Label Location):** Selects fields matching explicit row condition criteria or column names.


```python
# 1. Isolating structural target vectors
employee_matrix = df[['Employee', 'Salary_INR_K']]
print("--- Isolated Data Metrics ---")
print(employee_matrix)

# 2. Extracting coordinates using .iloc (Row 0, Col 1 -> 'Data Scientist')
print("\n--- Precision Coordinate Extraction ---")
print("Target Item via Index:", df.iloc[0, 1])

# 3. Filtering specific records based on logical values
high_earners = df[df['Salary_INR_K'] >= 90]
print("\n--- Filtered Rows (Salary >= 90k) ---")
print(high_earners)
```

### 4. File Management and Persistence
* Data files are loaded using `pd.read_csv()` and exported back onto your machine storage array via `.to_csv()`.


```python
# Exporting records onto the machine
df.to_csv("processed_it_workforce.csv", index=False)
print("Export Check: 'processed_it_workforce.csv' has been generated inside your directory!")

# Reading data from hard disk file
loaded_dataset = pd.read_csv("processed_it_workforce.csv")
print("\n--- Extracted Cities List ---")
print(loaded_dataset["Role"].unique())
```

##  MATHEMATICS WITH NUMPY

### 1. Ingestion Arrays and Shape Metrics



```python
import numpy as np

# 1. Instantiating 1D and 2D arrays
vector__1d = np.array([10, 20, 30, 40])
matrix_2d = np.array([[1, 2, 3], [4, 5, 6]])

print("--- Array Dimension Ingestion Diagnostics ---")
print(f"1D Array Shape Vector: {vector__1d.shape} | Dimensions: {vector__1d.ndim}")
print(f"2D Matrix Layout Shape: {matrix_2d.shape} | Dimensions: {matrix_2d.ndim}")
print(f"Data type tracking inside array: {vector__1d.dtype}")
```

### 2.  Vectorized Operations & Logic Arrays

```python
# Element-wise operations execute instantly without loop handlers
modified_vector = (vector__1d * 3) - 5

print("--- Array Arithmetic Vectorization ---")
print("Base Input Elements:", vector__1d)
print("Processed Output Vector Array ((x*3) - 5):", modified_vector)

# Advanced statistics computations
print("\n--- Rapid Global Statistical Computations ---")
print("Calculated Mean Average Metric:", vector__1d.mean())
print("Maximum Value Target Element:", vector__1d.max())
print("Sum across matrix columns (Axis=0):", matrix_2d.sum(axis=0))
```

### 3. Linear Algebra: Matrix Multiplications (Dot Product)



```python
# Defining mathematical coordinate structures
matrix_A = np.array([[1, 2], [3, 4]])
matrix_B = np.array([[5, 6], [7, 8]])

# Calculating dot cross matrix computations
transformation_matrix = np.dot(matrix_A, matrix_B)

print("--- Linear Algebra Evaluation (Matrix Dot Product) ---")
print("Matrix Product (A · B):\n", transformation_matrix)
```
