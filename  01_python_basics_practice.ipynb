## Module 1: Python Basics, Expressions & Strings

### 1. Variables and Data Types
* **Definition:** Variables are containers for storing data values.
* **Core Types:** 
  * `int`: Whole numbers (e.g., `10`, `-5`)
  * `float`: Decimals numbers (e.g., `3.14`, `10.0`)
  * `str`: Text wrapped in quotes (e.g., `"Hello"`)
  * `bool`: True or False values (`True`, `False`)
* **Type Casting:** Changing data from one type to another using functions like `int()`, `float()`, or `str()`.


```python

age = 20                 # int
pi_value = 3.1415        # float
name = "Sumon"           # str
is_student = True        # bool

print(type(age), type(pi_value), type(name), type(is_student))

string_num = "100"
converted_num = int(string_num)
print("Converted type:", type(converted_num))
```

### 2. Expressions and Mathematical Operations
* **Expressions:** Combinations of values, variables, and operators that evaluate to a single value.
* **Operators:** Standard math rules apply (PEMDAS).
  * `/` (Regular Division): Always returns a `float`.
  * `//` (Integer/Floor Division): Rounds down to the nearest whole integer.
  * `%` (Modulus): Returns the remainder of a division.


```python
result_standard = 11 / 3
result_floor = 11 // 3
result_remainder = 11 % 3

print("Standard Division (11/3):", result_standard)
print("Floor Division (11//3):", result_floor)
print("Remainder/Modulus (11%3):", result_remainder)
```

### 3. String Operations and Indexing
* **Indexing:** Accessing a single character using its position. Python starts counting at `0`. Negative indexing starts from `-1` (end of string).
* **Slicing:** Extracting a part of a string using `[start:stop]`. The `stop` index is excluded.
* **Methods:** Built-in string actions like `.upper()`, `.replace()`, and `.find()`.


```python
# String Operations
college = "IIT Madras"

# Indexing & Slicing
print("First letter:", college[0])
print("Last letter:", college[-1])
print("First three letters:", college[0:3])

# String Methods
print("Uppercase version:", college.upper())
print("Replace Madras with Chennai:", college.replace("Madras", "Chennai"))
print("Find index of 'Chennai':", college.find("Chennai"))
```

## Python Data Structures

### 1. Tuples
* **Characteristics:** Ordered, indexable, but **immutable** (cannot be changed, added to, or edited after creation).
* **Syntax:** Uses parentheses `(value1, value2)`.
* **Nesting:** Tuples can contain other tuples, lists, or strings inside them.


```python
# Creating a Tuple
my_tuple = ("Python", 3.11, [1, 2], ("nested", True))

print("Tuple length:", len(my_tuple))
print("Accessing nested tuple:", my_tuple[3][0])

# Proof of Immutability (the line below will crash Python)
# my_tuple[0] = "R" 
```

### 2. Lists
* **Characteristics:** Ordered, indexable, and **mutable** (can change elements, append new items, or delete existing ones).
* **Syntax:** Uses square brackets `[value1, value2]`.
* **Methods:** Common methods include `.append()`, `.extend()`, and `del`.


```python
# Creating and updating a List
skills = ["Python", "SQL", "Statistics"]

# Modifying elements
skills[2] = "Data Visualisation"

# Adding elements
skills.append("Machine Learning")  # Adds as 1 element
print("Updated Skills List:", skills)

# Slicing a list
print("Top 2 skills:", skills[0:2])
```

### 3. Dictionaries
* **Characteristics:** Unordered collections of **Key-Value pairs**. Highly efficient for looking up data.
* **Syntax:** Uses curly braces `{ "key": value }`.
* **Keys:** Must be unique and immutable (like strings or integers).


```python
# Creating a Data Science Student Profile Dictionary
student_profile = {
    "name": "Sumon",
    "year": 2,
    "degree": "Data Science",
    "completed_modules": ["Module 1", "Module 2"]
}

# Accessing values using keys
print("Student Name:", student_profile["name"])

# Adding a new key-value pair
student_profile["current_platform"] = "Coursera"

# Checking keys
print("All available fields:", student_profile.keys())
print("Full Profile:", student_profile)
```

### 4. Sets
* **Characteristics:** Unordered collections of **unique elements**. They automatically eliminate duplicates and do not support indexing.
* **Syntax:** Uses curly braces `{value1, value2}` or the `set()` function.
* **Mathematical Operations:** Used for powerful set operations like Union (`|`), Intersection (`&`), and Difference (`-`).


```python
# 1. Automatic duplicate removal
raw_user_ids = [101, 102, 101, 103, 102, 104]
unique_user_ids = set(raw_user_ids)
print("Original List:", raw_user_ids)
print("Unique Set (Duplicates Removed):", unique_user_ids)

# 2. Key Operations: Intersection and Union
python_students = {"Sumon", "Amit", "Rahul"}
sql_students = {"Rahul", "Priya", "Sumon"}

# Intersection: Who is studying BOTH?
both_courses = python_students.intersection(sql_students) # Or: python_students & sql_students
print("Students in both courses:", both_courses)

# Union: Who is studying AT LEAST ONE?
all_students = python_students.union(sql_students) # Or: python_students | sql_students
print("All unique students combined:", all_students)
```

## 1. Loops and Control Flow

### Key Concepts
* **For Loops:** Best for iterating over an exact sequence (like a list, tuple, or a specific `range()`).
* **While Loops:** Keep running continuously as long as a specific condition remains true.
* **Loop Modifiers:**
  * `break`: Completely stops the loop and exits out of it.
  * `continue`: Skips the rest of the current iteration and jumps directly to the next loop step.

```python
# 1. Standard For Loop with range
print("--- Standard For Loop ---")
for i in range(1, 4):
    print(f"Loop count: {i}")

# 2. While Loop with a break condition
print("\n--- While Loop with Break ---")
counter = 5
while counter > 0:
    if counter == 2:
        print("Break triggered at counter = 2")
        break
    print(f"Counter is: {counter}")
    counter -= 1

# 3. For Loop with a continue modifier
print("\n--- Loop with Continue ---")
numbers = [10, 15, 20, 25]
for num in numbers:
    if num % 2 != 0:  # Skip odd numbers
        continue
    print(f"Even number found: {num}")
```

## 2. Functions in Python

### Key Concepts
* **Definition:** Named blocks of code designed to perform a specific task, initialized using the `def` keyword.
* **Scope:** Variables defined inside a function are **local** to that function. They cannot be used outside it.
* **Return vs Print:** The `return` keyword passes data back to the main program so it can be saved in a variable. The `print()` function only displays text visually.


```python
# 1. Function with arguments and a return value
def calculate_bmi(weight_kg, height_m):
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, 2)

# Testing the function and storing the result
my_bmi = calculate_bmi(70, 1.75)
print(f"Calculated BMI stored in variable: {my_bmi}")

# 2. Understanding Scope (This demonstrates local vs global variables)
global_message = "I am global"

def scope_test():
    local_message = "I am local"
    print(f"Inside function: {local_message}")
    print(f"Inside function: {global_message}")

scope_test()
# Uncommenting the line below will cause a NameError because local_message is hidden outside
# print(local_message) 
```

## 3. Object-Oriented Programming (OOP)

### Key Concepts
* **Class:** A template or blueprint that defines the structural data and actions available for an object.
* **Object:** A unique instance created out of a class blueprint.
* **Attributes:** Variables defined inside a class that track specific information about an object.
* **Methods:** Regular functions defined inside a class that dictate what actions the object can perform.
* **`__init__` Method:** The constructor function that automatically triggers and runs whenever you create a new object.


```python
# 1. Defining a clean Class blueprint
class DataProject:
    # The Constructor method to set up attributes
    def __init__(self, project_name, language):
        self.project_name = project_name    # Attribute 1
        self.language = language            # Attribute 2
        self.is_completed = False           # Attribute 3 (Default value)

    # A custom method to alter attributes
    def complete_project(self):
        self.is_completed = True
        print(f"Success: '{self.project_name}' has been marked complete!")

    # A custom method to display info
    def display_status(self):
        status = "Completed" if self.is_completed else "In Progress"
        print(f"Project: {self.project_name} | Language: {self.language} | Status: {status}")

# 2. Instantiating an Object from the Class
project_01 = DataProject("Sales Analysis", "Python")

# 3. Interacting with methods and attributes
project_01.display_status()       # Check starting status
project_01.complete_project()     # Run the completion action
project_01.display_status()       # Verify updated status
```
